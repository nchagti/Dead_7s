## Calculate Probabilities

```python
import csv
import math
from collections import Counter

full_bag = {
    "A": 9, "B": 2,  "C": 2,
    "D": 4, "E": 12, "F": 2,
    "G": 3, "H": 2,  "I": 9,
    "J": 1, "K": 1,  "L": 4,
    "M": 2, "N": 6,  "O": 8,
    "P": 2, "Q": 1,  "R": 6,
    "S": 4, "T": 6,  "U": 4,
    "V": 2, "W": 2,  "X": 1,
    "Y": 2, "Z": 1,
}

def falling_factorial(n: int, k: int) -> int:
    """Compute n * (n-1) * ... * (n-k+1)."""
    prod = 1
    for x in range(n, n - k, -1):
        prod *= x
    return prod

def score_word(word: str) -> int:
    """Number of ways to draw letters for word from bag."""
    counts = Counter(word)
    prod = 1
    for letter, v in counts.items():
        n = full_bag[letter]  # assumes letter always exists in full_bag
        numerator = falling_factorial(n, v)
        denom = math.factorial(v)  # same as gamma(v+1)
        prod *= numerator // denom  # exact integer division here
    return int(prod)


# Compute, then sort by (-score, word)
probabilities = [(w, score_word(w)) for w in sorted_dead_sevens]
probabilities.sort(key=lambda t: (-t[1], t[0]))

# Write CSV (three columns: word, probability order, ways to draw)
out_path = r'.\dead_7s_by_probability.csv'
with open(out_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Alphagram", "Probability", "Bag Draw Combo"])

    # rows
    for rank, (alphagram, combo_count) in enumerate(probabilities, start=1):
        writer.writerow([alphagram, rank, combo_count])
```
